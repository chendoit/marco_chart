```python
import os
import pickle
import pandas as pd
# import matplotlib.pyplot as plt
from ipychart import Chart
```

```python
# 設置數據目錄的路徑
DATA_DIR = r"C:\code\youtube\2025-01-11-MacroMicro-Data\flask_charts\data"

# 載入數據的函數
def load_series_data(chart_id):
    """Load series data from the pickle file."""
    file_path = os.path.join(DATA_DIR, f"series_{chart_id}.pkl")  # 使用正確的路徑
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"File series_{chart_id}.pkl not found in {DATA_DIR}")
    with open(file_path, 'rb') as f:
        return pickle.load(f)
```

```python
def generate_chart_data(chart_id_list, months=6):
    """Generate chart data for Chart.js."""
    chart_data = {"labels": [], "datasets": []}

    # 遍歷所有 chart_id
    for chart_id in chart_id_list:
        # 載入數據
        data = load_series_data(chart_id)
        series_data = data['data']

        # 轉換為 DataFrame
        df = pd.DataFrame(series_data, columns=['Date', 'Value'])
        df['Date'] = pd.to_datetime(df['Date'])
        df['Value'] = df['Value'].astype(float)

        # 過濾最近 N 個月數據
        max_date = df['Date'].max()
        min_date = max_date - pd.DateOffset(months=months)
        df = df[df['Date'] >= min_date]

        # 更新 Chart.js 格式的數據
        if not chart_data["labels"]:
            chart_data["labels"] = df['Date'].dt.strftime('%Y-%m-%d').tolist()
        chart_data["datasets"].append({
            "label": data['title'],
            "data": df['Value'].tolist(),
            "borderColor": "rgba(75, 192, 192, 1)",
            "backgroundColor": "rgba(75, 192, 192, 0.2)",
            "fill": False
        })

    return chart_data
```

```python
# 測試和繪製圖表
if __name__ == "__main__":
    chart_id_list = [4934, 8219]  # 替換為你的 chart_id
    chart_data = generate_chart_data(chart_id_list)


    # 使用 ipychart 繪製圖表
    chart = Chart(
        data=chart_data,
        kind='line',
        options={
            'scales': {
                'y': {
                    'beginAtZero': True
                }
            }
        }
    )
    chart
```
