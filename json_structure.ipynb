```python
import json
from collections import defaultdict

def describe_json_structure(data, indent=0):
    description = ""
    if isinstance(data, dict):
        description += " " * indent + "這是一個字典，包含以下鍵：\n"
        for key, value in data.items():
            description += " " * (indent + 2) + f"鍵 '{key}' 對應的值是：\n"
            description += describe_json_structure(value, indent + 4)
    elif isinstance(data, list):
        description += " " * indent + "這是一個列表，包含以下元素：\n"
        if len(data) > 0:
            description += describe_json_structure(data[0], indent + 2)
        else:
            description += " " * (indent + 2) + "列表為空\n"
    else:
        description += " " * indent + f"這是一個 {type(data).__name__} 類型的值\n"
    return description

# 示例 JSON
json_data = '''
{
    "name": "John",
    "age": 30,
    "address": {
        "street": "123 Main St",
        "city": "New York"
    },
    "phoneNumbers": [
        {"type": "home", "number": "212 555-1234"},
        {"type": "office", "number": "646 555-4567"}
    ]
}
'''

# 解析 JSON
data = json.loads(json_data)

# 描述 JSON 結構
description = describe_json_structure(data)
print(description)
```

```python
import os, pickle
def load_series_data(chart_id):
    """Load series data from the pickle file."""
    file_path = os.path.join(r"C:\code\youtube\2025-01-11-MacroMicro-Data\data", f"series_{chart_id}.pkl")
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"File series_{chart_id}.pkl not found in ./data")
    with open(file_path, 'rb') as f:
        return pickle.load(f)
```
